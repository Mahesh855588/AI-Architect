DECORATORS

```python
# decorator is function taking function as paramenter and returns function
# manual
def my_decorator(func):
    def wrapper():
        print("start")
        func()
        print("end")
    return wrapper

# without changing the body of greet() method we can do some actions with decorators
def greet():   
    print("hello mahesh")

g=my_decorator(greet) 
g()

# decorator
@my_decorator
def greeting():   # by calling the method we can acces the decorator 
    print("hello asritha")
greeting()
```

```python
# handling arguments
def my_decorator(func):
    def wrapper(*args,**kwArgs):
        print("before")
        result=func(*args,**kwArgs)
        print("after")
        return result # None if you forget the return
    return wrapper

@my_decorator
def Add(a,b,c):
    return a+b+c

Add(1,2,3)
print(Add.__name__) # returns wrapper
print(Add.__doc__) # returns none
# to reach until original better to use @wraps(func)

from functools import wraps
def my_decorator(func):
    @wraps(func)
    def wrapper(*args,**kwArgs):
        print("before")
        result=func(*args,**kwArgs)
        print("after")
        return result # None if you forget the return
    return wrapper

@my_decorator
def Add(a,b,c):
    ''' adding'''
    return a+b+c

Add(1,2,3)
print(Add.__name__) # returns wrapper
print(Add.__doc__) # returns none
```

```python
# decorators with arguments
from functools import wraps
def repeat(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args,**kwArgs):
            for i in range(times):
                result=func(*args,**kwArgs)
            return result
        return wrapper
    return decorator

@repeat(4)
def greet(name):
    print(f"Hello, {name}")  

greet("mahesh")
```
