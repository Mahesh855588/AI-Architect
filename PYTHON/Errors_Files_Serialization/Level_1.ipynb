```python
# try and except

try:
    print(10/0)
except ZeroDivisionError:
    print("cannot divide by zero")
```

```python
# else
# entire block executes successfully else executes

try:
    print(10/0)
except ZeroDivisionError:
    print("cannot divide by zero")
else:
    print("block executed successfully")

print("---------")

try:
    print(10/10)
except ZeroDivisionError:
    print("cannot divide by zero")
else:
    print("block executed successfully")
```

```python
# catching the exception
try:
    print(10/0)
except ZeroDivisionError as e: #catching with e
    print(f"error is : {e}")


# multiple exceptions
try:
    print(int("abc"))
except ValueError as e:
    print("not a valid number")
except ZeroDivisionError as e:
    print("divide by zero")
except TypeError as e:
    print("wrong type")

# one block for several types
try:
    print(int("abc"))
except (ValueError,ZeroDivisionError,TypeError) as e:
    print(f"error is : {e}")
```

```python
# finally 
# always runs even exception occurs

try:
    print(10/0)
except  ZeroDivisionError as e:
    print(f"error is : {e}")
finally:
    print("i always runs")
```

```python
# exception hierarchy


'''
BaseException
 └── Exception
      ├── ValueError
      ├── TypeError
      ├── KeyError
      ├── IndexError
      ├── AttributeError
      ├── ZeroDivisionError
      ├── FileNotFoundError
      ├── OSError
      │    └── (FileNotFoundError is actually here too)
      └── ArithmeticError
           └── ZeroDivisionError (again — some sit under multiple parents)
'''

# order matters 
# before type exception if we keep base execption always it will execute because it was the parent(is a)
try:
    print(10/0)
except Exception as e:
    print("Hi ")
except ZeroDivisionError as e:
    print(e) # but we need this 


# this is proper order
try:
    print(10/0)
except ZeroDivisionError as e:
    print(e) # 
except Exception as e:
    print("Hi ")
```

```python
# custom exceptions

class InsufficientFunds(Exception): # must inherit exception
    pass

def withdraw(amount,balance):
    if amount>balance:
        raise InsufficientFunds("Insufficient funds please check")

try:
    withdraw(500,100)
except InsufficientFunds as e:
    print(f"transaction failed {e}")
```
