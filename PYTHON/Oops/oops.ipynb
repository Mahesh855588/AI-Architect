```python
# Bank account example
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
    def deposit(self,amount):
        self.balance += amount
        return self.balance
    def withdraw(self,amount):
        if(amount>self.balance):
            return "Insufficient balance"
        self.balance-=amount
        print(f"Remaining balance is {self.balance}")
acc1=BankAccount("Mahesh",50000)
print(acc1.deposit(4000))
print(acc1.withdraw(90000))
print(acc1.withdraw(9000))
```

```python
# Inheritance
# should have HAS-A relation
# creating a new class with the help of existing one
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print(f"{self.name} speaks!")

class Dog(Animal): # dog can access everything from animal
    pass
d=Dog("Rex")
d.speak()
```

```python
# overriding
# same function name as parent with different behavior
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print(f"{self.name} speaks")
class Dog(Animal):
    def speak(self):
        print(f"{self.name} speaks loudly")

d=Dog("Rex")
d.speak()
```

```python
# super()
# extending children behaviour from parent
# calling parent
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print(f"{self.name}")
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    def speak(self):
        super().speak()
        print(f"{self.breed} is the breed")
d=Dog("Rex","Labrador")
d.speak()
```

```python
# inheritance example
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def details(self):
        print(f"{self.name}: {self.salary}")
class Manager(Employee):
    def __init__(self,name,salary,team_size):
        super().__init__(name,salary)
        self.team_size=team_size
    def details(self):
        super().details()
        print(f"{self.name} manager has {self.team_size} team size")

m=Manager("mahesh",8000,5)
print(m.name,m.salary,m.team_size)
```

```python
# multiple inheritance
# more than 1 parent for 1 child
class A:
    def doingA(self):
        print("does A")
class B:
    def doingB(self):
        print("does B")
class C(A,B):
    pass
c=C()
c.doingA()
c.doingB()
print(C.__mro__)
```

Inheritance in python follows MRO(method resolution order) as follows

```python
# diamond problem for MRO
class A:
    def greet(self):
        print("A")
class B(A):
    def greet(self):
        print("B")
class C(A):
    def greet(self):
        print("C")
class D(B,C):
    pass
d=D()
d.greet()
print(D.__mro__)
```

```python
# Mixins
import json
class JsonMixings:
    def to_json(self):
        return json.dumps(self.__dict__)
class CompatableMixings:
    def __eq__(self,other):
        return self.__dict__==other.__dict__
class Person(JsonMixings,CompatableMixings):
    def __init__(self,name,age):
        self.name=name
        self.age=age
p=Person("Mahesh",25)
q=Person("Asritha",25)
print(p.to_json())
print(p.__eq__(p))
print(p.__eq__(q))
```

```python
# @property

# problem
class Person:
    def __init__(self,age):
        self.age=age
    def print_age(self):
        print(f"{self.age}")
p=Person(25)
p.print_age()
p.age=35
p.print_age()
p.age=-5
p.print_age() # returns -5 which should not happen
# when assigning to age no code runs
# to overcome this problem we should run some code to validate the assignment
# @property (getter and setter)
class Person:
    def __init__(self,age):
        self.age=age
    @property # getter
    def age(self):
        return self._age
    
    @age.setter
    def age(self,value):
        if value<0:
            return ValueError("age cannot be negative")
        self._age=value
# from cretaion of instance of object setter starts working
p=Person(9)
print(p.age)
```

```python
# DUnders

# __init()
# this will be called when we create an instance
class Person:
    def __init__(self,name):
        self.name=name
p=Person("Mahesh") # here __init__() method will be called automatically
print(p.name)


# __str__()
# this will be called when we call print()
class Person:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return f"this is {self.name}!"
p=Person("Mahesh")
print(p)


# __repr__()
# for printing list,dict,set,tuple of objects by default uses __repr__()
# this is for developer view
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f"this is {self.name} and {self.age} years old"
    def __repr__(self):
        return f"Person(\"{self.name}\",{self.age})"
p=Person("Mahesh",25)
print(p)
print(repr(p))
print([p]) 



# __eq__()
# when == used in print() __eq__() will be called
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __eq__(self,other):
        return self.name==other.name and self.age==other.age
p1=Person("mahesh",25)
p2=Person("mahesh",25)
p3=Person("mahesh",26)
print(p1==p2) # calls __eq__()
print(p1==p3) # calls __eq__()


# __hash__()
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __eq__(self,other):
        return self.name==other.name and self.age==other.age
    def __hash__(self):
        return hash(self.name),hash(self.age)
p1=Person("mahesh",25)
p2=Person("mahesh",25)
print(p1==p2) # if p1 and p2 are equal in set it should store 1 object but it stores both
# whatever changed in __eq__() make it in __hash__()


# __len__()
# when len() is called
class PlayList:
    def __init__(self,songs):
        self.songs=songs
    def __len__(self):  #without this method throws error object of type 'Playlist' has no len()
        return len(self.songs)

p=PlayList([1,2,3,4,5])
print(len(p))


# __getitem__(),__setitem__()
class PlayList:
    def __init__(self,songs):
        self.songs=songs
    def __len__(self):
        return len(self.songs)
    def __getitem__(self,index):
        return self.songs[index]
    def __setitem__(self,index,value):
        self.songs[index]=value
    def __repr__(self):
        return f"playlist is {self.songs}"
    def __delitem__(self,index):
        del self.songs[index]

p=PlayList([1,2,3,4,5,6])
print(len(p))
print(p[0])
p[0]=9
print(p[0])
print(p)
del p[0]
print(p)

# __iter__()
# for loop working on an object
class PlayList:
    def __init__(self,songs):
        self.songs=songs
    def __iter__(self):
        return iter(self.songs)
p=PlayList([1,2,3,4,5,6])
for i in p:
    print(i)


# __call__()
# calling an object like function
class Person:
    def __init__(self,greeting):
        self.greeting=greeting
    def __call__(self,name):
        return f"{self.greeting} {name}"
p=Person("Hello")
print(p("mahesh"))

# operator overloading
# arithmetic operations on objects
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
v1=Vector(1,2)
v2=Vector(3,4)
print(v1+v2)
 
'''__add__        +
__sub__        -
__mul__        *
__truediv__    /
__floordiv__   //
__mod__        %
__pow__        **
'''
```

```python
# Data classes

# normal
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __repr__(self):
        return f"this is {self.name} and {self.age} years old"
    def __eq__(self,other):
        return (self.name,self.age)==(other.name,other.age)


# data class
from dataclasses import dataclass
@dataclass
class Person:
    name:str
    age:int
    city:str
# That's it. You get all of the above __init__, __repr__, and __eq__

p=Person("Mahesh",25,"bengaluru")
print(p)
print(p==Person("Mahesh",25,"bengaluru"))
```

```python
# __slots__()
# every object carries dictionary to its attributes
# with slots you will eliminate dictionaries
# this is usefull when creating millions of instances

class Person_with_slots:
    __slots__="name","age"
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Person_with_dict:
    def __init__(self,name,age):
        self.name=name
        self.age=age
a=Person_with_dict("mahesh",26)
b=Person_with_slots("mahesh",26)
import sys
print(sys.getsizeof(a)+sys.getsizeof(a.__dict__))
print(sys.getsizeof(b)) # here we are removing dictionary
# cannot add new attributes dynamically

# creating slots with data class
@dataclass(slots=True)
class Person:
    name:str
    age:int
    city:str
p=Person("mahesh",25,"bengaluru")
print(p)
p.name="asritha"
print(p)
#p["school"]="zphs" #'Person' object does not support item assignment
```

```python
# ABC(abstract base class)
# every method in base class should be overriden
# base class inherits ABC
from abc import abstractmethod,ABC
class Shape(ABC):
    @abstractmethod
    def area(self):  #abstract method
        pass
    def details(self): #normal method
        return self.area()
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
c=Circle(5)
print(c.area())
print(c.details())

# with property

from abc import abstractmethod,ABC
class Shape(ABC):
    @property
    @abstractmethod
    def area(self):  #abstract method
        pass
    def details(self): #normal method
        return self.area()
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    @property
    def area(self):
        return 3.14*self.radius**2
c=Circle(5)
print(c.area)
```

```python
# inheritance (IS-A relation)
# composition (Has a relation)
class Engine:
    def start(self):
        return "engine started"
class Car:
    def __init__(self):
        self.engine=Engine() # has a relation 
    def start(self):
        return self.engine.start()

c=Car()
print(c.start())
```
