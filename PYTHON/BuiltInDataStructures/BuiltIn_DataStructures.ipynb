LISTS

```python
#creation of lists
a=[1,2,3,4]
b=[1,"two",3,"four"]
c=[]
print(a)
print(b)
print(c)
```

```python
#length of list
len(a)
```

```python
#Adding
a=[1,2,3]
a.append(5) # Add to end ->O(1)
print(a)
a.insert(0,2) # Insert at position ->O(n) # Insert at 0th index # shifts everything right
print(a)
a.insert(1,2) # Insert at 1st position
print(a)
```

```python
#Removing and common methods
a=[2,6,2,8,3,4,6,9,3]
a.pop() # remove and return last element ->O(1)
print(a)
a.pop(0) #remove and return 1st element ->O(n)
print(a)
a.pop(1) #remove and return 2nd element ->O(n)
print(a)
a.remove(4) # remove by value(first match)
print(a)
del a[1] # remove 1st index element,no return value
print(a)
a=[2,6,2,8,3,4,6,9,3]
a.sort() # ascending order
print(a)
a.sort(reverse=True) # decending order
print(a)
a.reverse() # reverse in place
print(a)
print(a.count(2)) # occurences of value
print(a.index(6)) # index of 1st occurence value
a.extend([9,8,7,6,5])
print(a) # appends multiple items at once
a.clear() # emptying list
print(a)
```

```python
# + vs extend() vs append()
a = [1, 2]
print(a)
a.append([3, 4])    # [1, 2, [3, 4]] — adds the list AS ONE ELEMENT
print(a)
a.extend([3, 4])     # [1, 2, 3, 4]  — adds each element individually
print(a)
b = a + [5, 6]         # creates a NEW list, doesn't modify a
print(b)
```

```python
#Slicing
a=[1,2,3,4,5,6,7,8,9]
print(a)
# starts from 0th index
b=a[:5] # slicing until 5 indexes(excluding last)
print(b)
c=a[0:6] # slicing until 6 indexes
print(c)
d=a[5:] # slicing from 5th index
print(d)
e=a[2:6] # slicing from and to
print(e)
f=a[::-1] # reversing list
print(f)
g=a[::2] # every 2nd step from index 0
print(g)
```

```python
# sort() vs sorted()
a=[1,2,3,4]
a.sort(reverse=True) # reverse in place and returns None
new_List=sorted(a) # returns new sorted list and original will be unaffected
print(new_List)
print(a)
```

```python
# copying list
a=[1,2,3,4,[1,2,3]]
b=a # here if b changes a will change as this is mutable
b.append(5)
print(a)
print(b)
c=a.copy() # only nested elements will be shared among remaining will be independent(shallow copy)
a=a[:] # another way of shallow copy
c.pop()
c[4].append(5)
print(a)
print(c)
import copy
d=copy.deepcopy(a) # here everything is independent including nested elements
d[4].append(8)
d.pop()
print(a)
print(d)
```

```python
#List as a stack and queue
# Stack (LIFO) — use append/pop, both O(1)
stack = []
stack.append(1)
stack.append(2)
stack.pop()      # removes 2
```

```python
# Queue (FIFO) — pop(0) is O(n), inefficient for large lists
from collections import deque
queue = deque()
queue.append(1)
queue.popleft()   # O(1) — proper way to do queues in Python
```

```python
#sorting with key
people = [("puri", 24), ("Mahesh", 31)]
people.sort(key=lambda p: p[1])          # sort by age
print(people)
words = ["banana", "kiwi", "apple"]
sorted(words, key=len)                     # sort by length, returns NEW list
# ['kiwi', 'apple', 'banana']
```

TUPLES

```python
# creating tuples
a=(1,2,3,4,5)
b=(2,) # tuple should be comma separated without it will be int with parenthesis
empty_Tuple=() # This is empty tuple
c=1,2,3,4,5,6  # parenthesis are actually optional
print(a)
print(b)
print(c)
```

```python
# Tuples are immutable
# Once tuples created cannot be changed after that
a=(1,2,3,4,5)
#a[1]=100 #Type error

# But mutable types(lists,sets,dicts) inside tuples are immutable
a=(1,2,3,4,5,[6,7,8])
a[5].append(9)
print(a)
```

```python
#Indexing and slicing — same as lists
t = (10, 20, 30, 40)
print(t[0])       # 10
print(t[-1])      # 40
print(t[1:3])     # (20, 30)
print(t[::-1])    # (40, 30, 20, 10)
```

```python
coords = {(0, 0): "origin", (1, 1): "diagonal"}  # ✅ tuple as dict key
#coords2 = {[0, 0]: "origin"}  # ❌ TypeError — list isn't hashable
```

```python
#Tuple methods
t = (1, 2, 2, 3)
print(t.count(2))   # 2
print(t.index(3))   # 3
#That's it — no append, remove, sort, etc., since those would require mutation.
```

```python
# tuple unpacking
point=(2,3)
x,y=point
print(x,y)
```

```python
# swapping values 
# classic python idiom using tuples
a,b=1,2
a,b=b,a
print(a,b)
```

```python
# unpacking with *
a=(1,2,3,4,5,6)
first,*middle,last=a
print(first)
print(middle)
print(last)
```

```python
# Nested Tuples
a=((1,2),(3,4))
print(a[0][1])
```

```python
# converting between list and tuple
t=(1,2,3)
l=list(t) # converting tuple to list
b=tuple(l) # converting list back to tuple
print(t,l,b)
```

```python
# named tuples
from collections import namedtuple
Point=namedtuple("Point",["x","y"])
p=Point(1,2)
print(p)
```

```python
# tuple vs List
import sys
print(sys.getsizeof([1,2,3])) # larger
print(sys.getsizeof((1,2,3))) # smaller
```

```python
# tuple as function return value
def Min_Max(numbers):
    return min(numbers),max(numbers)
low,high=Min_Max([1,2,3,4,5,6,7,8,9])
print(low,high)
```

```python
# Tuple concatenation
t1=(1,2)
t2=(3,4)
print(t1+t2)
# Tuple multiplication
t3=t1*5
print(t3)
```

```python
#Comparing tuples
print((1, 2, 3) < (1, 2, 4))   # True — compares element by element
print((1, 2) < (1, 2, 3))       # True — shorter tuple is "less" if it's a prefix
```

Dictionaries

```python
# Creating dictionaries
d={"Mahesh":24,"Puri":25} #Default declaration
d1={} #empty dictionary
d2=dict(name="Mahesh",age=25,color="Blue") # constructor declaration
d3=dict([("name","mahesh"),("age",24),("color","blue")]) # from list of tuples
print(d)
print(d1)
print(d2)
print(d3)
```

```python
# accesing dictionary
d=dict(name="Mahesh",age=25,color="Blue")
print(d)
print(d["age"]) # if key is not present in dictionary it will throw error
# best way to access is with .get()
# if key is not present it will return None
print(d.get("name"))
print(d.get("Name"))
print(d.get("color",-1)) # returns -1 if key is not present
```

```python
# Adding and Updating
d["school"]="ZPHS"
print(d)

d["name"]="Jinka Mahesh Reddy"
print(d)
```

```python
# Remove items form dictionary
d=dict(name="Mahesh",age=25,color="Blue",school="ZPHS")
del d["school"] # removes but doesn't return anything
print(d)
d.pop("age") # removes and returns value that deleted
print(d)
d.clear() # clearing entire dictionary
print(d)
```

```python
# checking if key exists
d=dict(name="Mahesh",age=25,color="Blue",school="ZPHS")
print("name" in d)
print("college" in d)
```

```python
# Iterating over dictionaries
d=dict(name="Mahesh",age=25,color="Blue",school="ZPHS")
for i in d:
    print(i) # prints keys
print("\n")

for i in d.keys():
    print(i) # returns same as above
print("\n")

for i in d.values():
    print(i)
print("\n")

for i,j in d.items():
    print(i,j)
    
```

```python
# nested dictionaries
d={"Vowels":{1:"a",2:"e",3:"i",4:"0",5:"u"},"Consonents":{1:"b",2:"c",3:"d",4:"f"}}
print(d["Vowels"][1])
```

```python
# merging dictionaries
m1=dict(a=1,b=2,c=3)
m2=dict(d=4,e=5,f=6)
m3=m1|m2
print(m1)
print(m2)
print(m3)

m1.update(m2) # m1 will be updated and m2 will stay remain
print(m1)
print(m2)
print(m3)
```

```python
#Dictionary keys must be hashable(Immutable)
d={(1,2):"coords1"} # correct tuple is immutable
 #d1={[1,2]:"coords2"} # incorrect list is mutable # throws error
print(d)
```

```python
# set default
m1=dict(a=1,b=2,c=3)
m1.setdefault("d",90)
print(m1)
#if m1 containd d value it will not override else it will create new value
m1.setdefault("d",10)
print(m1)
```

```python
# default dict
# if we add values to the keys if key is not present it will throw error
# defaultdict will handle this without throwing error
from collections import defaultdict
d=defaultdict(int)
d["a"]+=1
print(d)
d["a"]+=1
print(d)
d["a"]+=2
d["c"]+=67
d["b"]+=3
print(d)
```

```python
# counter
from collections import Counter
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts=Counter(words)
print(counts)
```

```python
# ordered dictionary
from collections import OrderedDict
d1 = {"a": 1, "b": 2}
d2 = {"b": 2, "a": 1}
print(d1 == d2)   # True — plain dicts ignore order when comparing

od1 = OrderedDict([("a", 1), ("b", 2)])
od2 = OrderedDict([("b", 2), ("a", 1)])
print(od1 == od2)   # False — OrderedDict compares order too

od = OrderedDict([("a", 1), ("b", 2), ("c", 3)])
od.move_to_end("a")
print(od)   # OrderedDict([('b', 2), ('c', 3), ('a', 1)])

od.move_to_end("c", last=False)   # move to the front instead
```
