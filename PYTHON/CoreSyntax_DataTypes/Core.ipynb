```python
#Variables
# variables are just tags(sticy note) that pointing at a box somewhere in memory
x=100
y="Mahesh"
z=10.78
print(x,y,z)
```

```python
#Equality(==)
# "==" whether both variables have same values
a=10
b=10
c=18
#here a and b points to different memory objects in memory
print(a==b)
print(a==c)
```

```python
#Identity(is)
# "is" whether both variables points to same memory location
a=[1,2,3]
b=[1,2,3]
c=a
#here a and b have same values but points to different memory locations
# c shares same memory as a
print(a is b)
print(a is c)
```

```python
#Mutability
# value can be changed in place(will let you overwrite)
# list,set,dict
a=[1,2,3]
print(a)
print(id(a))
a.append(4) # 4 will be added in place to same memory box
print(a)
print(id(a))
```

```python
#Immutability
# value cannot be changed in place(will not let you overwrite)
# int,str,float,tuple,bool
a=10
print(a)
print(id(a))
a=50
print(a)
print(id(a))
# here when we assign 50 it will create new memory and points to it
b="Hello"
print(b)
print(id(b))
b="Hello world"
print(b)
print(id(b))
```

```python
#f-strings (modern way to format variables into text)
name="Mahesh"
age=12
print(f"My name is {name} and I am {age} years old.") # prints "My name is Mahesh and I am 12 years old."
```

```python
#f-strings
name = "Mahesh"
score = 92.567

print(f"Name: {name}")
print(f"Score: {score:.2f}")     # 2 decimal places → 92.57
print(f"{name.upper()}")          # can call methods inside {}
print(f"{score:>10}")             # right-align in 10-char width
print(f"{score:,.2f}")            # comma separators for large numbers with 2 deciamals
pct = 0.4567
print(f"{pct:.1%}")                          # 45.7%  -- percent formatting built in
'''Breaking down {pct:.1%}

The format spec after the : is .1%. Two parts:

% — percentage type: This tells Python to do two things automatically:
Multiply the value by 100
Append a % symbol
.1 — precision: Number of digits after the decimal point (in the percentage form, not the original number).'''
```

```python
#Creating strings
s1 = 'hello'
s2 = "hello"
s3 = '''This is mahesh
I am 18 years old'''
s4 = """this is puri
she is 81 years old"""
print(s1)
print(s2)
print(s3)
print(s4)
```

```python
#Indexing and Slicing
s = "Hello, World"

print(s[0])      # 'H'
print(s[-1])     # 'd' — negative indexing from the end
print(s[0:5])    # 'Hello' — slice [start:end), end excluded
print(s[:5])     # 'Hello' — omit start = from beginning
print(s[7:])     # 'World' — omit end = to the end
print(s[::-1])   # 'dlroW ,olleH' — reverse the string
print(s[::2])    # 'Hlo ol' — step of 2
```

```python
#Strings are immutable
#for immutable inplace assignment is not possible
s = "hello"
print(s)
print(id(s))
#s[0] = "H"   # TypeError — can't modify a string in place
s = "Hello"  # this creates a NEW string, doesn't modify the old one
print(s)
print(id(s))
```

```python
#Common string methods
s = "  Hello World  "

print(s.strip())        # "Hello World" — removes leading/trailing whitespace
print(s.lower())        # "  hello world  "
print(s.upper())        # "  HELLO WORLD  "
print(s.replace("World", "Python"))  # "  Hello Python  "
print(s.split())        # ['Hello', 'World'] — splits on whitespace by default
print("-".join(["a", "b", "c"]))  # "a-b-c"
print(s.strip().startswith("Hello"))  # True
print(s.strip().endswith("World"))    # True
print(s.find("World"))   # index where found, or -1 if not found
print(s.count("l"))      # counts occurrences
```

```python
#String immutability performance note
# Inefficient — creates a new string object on every iteration
result = ""
for i in range(1000):
    result += str(i)
print(result)
# Better — join is more efficient for many concatenations
result = "".join(str(i) for i in range(1000))
print(result)
```

```python
#Encoding/decoding basics
s = "hello"
encoded = s.encode("utf-8")     # b'hello' — bytes object
decoded = encoded.decode("utf-8")  # back to "hello"
text = "café"
encoded = text.encode("utf-8")     # str -> bytes
print(encoded)                      # b'caf\xc3\xa9'
print(type(encoded))                # <class 'bytes'>

decoded = encoded.decode("utf-8")  # bytes -> str
print(decoded)                      # café

b = b"hello"          # byte string literal
print(type(b))         # <class 'bytes'>
print(b + text.encode())  # concatenation works only bytes-to-bytes
```

OPERATORS

```python
#Arithmetic Operators
a, b = 10, 3

print(a + b)   # 13  addition
print(a - b)   # 7   subtraction
print(a * b)   # 30  multiplication
print(a / b)   # 3.333...  true division (always float)
print(a // b)  # 3   floor division
print(a % b)   # 1   modulo (remainder)
print(a ** b)  # 1000  exponentiation
```

```python
#Comparison Operators
print(a == b)  # False — equal to
print(a != b)  # True  — not equal to
print(a > b)   # True
print(a < b)   # False
print(a >= b)  # True
print(a <= b)  # False
```

```python
#Logical Operators
x = True
y = False

print(x and y)  # False — both must be True
print(x or y)   # True  — at least one True
print(not x)    # False — flips the value
```

```python
print(3 and 5)      # 5    -- both truthy, returns the LAST evaluated operand
print(0 and 5)       # 0    -- short-circuits at first falsy value, returns it
print(3 or 5)        # 3    -- short-circuits at first truthy value, returns it
print(0 or 5)         # 5    -- 0 is falsy, so evaluation continues to 5
```

```python
#Assignment Operators
x = 5
print(x)
x += 3   # x = x + 3  → 8
print(x)
x -= 2   # x = x - 2  → 6
print(x)
x *= 2   # x = x * 2  → 12
print(x)
x /= 4   # x = x / 4  → 3.0
print(x)
x //= 1  # floor divide in place
print(x)
x **= 2  # power in place
print(x)
x %= 5   # modulo in place
print(x)
```

```python
#Bitwise Operators
a, b = 5, 3   # binary: 101, 011

print(a & b)   # 1   AND
print(a | b)   # 7   OR
print(a ^ b)   # 6   XOR
print(~a)      # -6  NOT (inverts all bits)
print(a << 1)  # 10  left shift (multiply by 2)
print(a >> 1)  # 2   right shift (divide by 2)
```

```python
#Identity Operators
a = [1, 2]
b = [1, 2]
c = a

print(a is b)      # False — different objects, same value
print(a is c)      # True  — same object
print(a is not b)  # True
```

```python
#Membership Operators
nums = [1, 2, 3]
print(2 in nums)      # True
print(5 not in nums)  # True
```

```python
#Operator precedence
print(2 + 3 * 4)      # 14, not 20 — * before +
print((2 + 3) * 4)    # 20 — parentheses override
#Rough order (high to low): ** → unary -/+ → *, /, //, % → +, - → comparisons → not → and → or
```

```python
#Chained comparisons
x = 5
print(1 < x < 10)          # True — same as (1 < x) and (x < 10)
print(1 < x < 10 < 20)     # True — chains any length
```

```python
#Walrus operator
# Normal way
n = 10
if n > 5:
    print(n)

# Walrus way — assign AND use in one expression
if (n := 10) > 5:
    print(n)  # 10
```

```python
#in / not in
print("cat" in "concatenate")  # True — substring check
```

```python
#Ternary-style
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
```

```python
#Augmented assignment on mutable objects — subtle trap
a = [1, 2]
b = a
a += [3]      # this MODIFIES the list in place
print(b)      # [1, 2, 3] — b changed too!
```

```python
a = [1, 2]
b = a
a = a + [3]   # this creates a NEW list
print(b)      # [1, 2] — b unaffected
```

```python
# For loop
for i in range(10):
    print(i)

fruits=["apple","banana","guava"]
for fruit in fruits:
    print(fruit)
```

```python
#  Range() variations
print(list(range(5))) #0,1,2,3,4
print(list(range(2,5))) # 2,3,4
print(list(range(0,20,2))) #(start,stop,step)
print(list(range(0,10,-1))) #negative step from last to first
```

```python
# while 
#until condition mets it will loop
count=0
while count<=5:
    print(count)
    count+=1
```

```python
#Break 
# Exit the loop immmediately
for i in range(10):
    if(i==5):
        break
print(f"break {i}")
```

```python
#continue 
# skip to next loop
for i in range(10):
    if(i%2==0):
        continue
    else:
        print(i)
```

```python
# else on loops
for i in range(5):
    print(i)
else:
    print("loop executed succefully")

#if loop breaks in between skips else loop
for i in range(5):
    if i==3:
        break
else:
    print("loop executed successfully without breaking")
```

```python
# Nested loops
for i in range(5):
    for j in range(5):
        print(i,j)
```

```python
# enumerate()
#along with items it will return the position of item
vowels=['a','e','i','o','u']
for i,x in enumerate(vowels):
    print(i,x)
```

```python
# zip()
# if we have 2 related lists that goes together then we use zip()

#Normal way
names=["Jinka","mahesh","reddy"]
marks=[90,90,99]
for i in range(len(names)):
    print(names[i],marks[i])

#zip()
for i,j in zip(names,marks):
    print(i,j)
```

```python
# Dictionaries
d=dict(a=1,b=2,c=3,d=4)
for i,j in d.items():
    print(i,j)
```

```python
# else with while
# if loop completes without breaking it will execute else loop similar as for loop with else
count=0
while count<=5:
    print(count)
    count+=1
else:
    print("loop completed successfully")
```

```python
#Reversed iteration with reverse()
for i in reversed(range(10)):
    print(i)
```

```python
#Match
# This is similar to switch case
day=int(input())
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("wednesday")
    case 4:
        print("thursday")
    case 5:
        print("friday")
    case 6:
        print("saturday")
    case 7:
        print("sunday")
    case _:
        print("enter correct input")
```

```python
# multiple values in one case
day=int(input())
match day:
    case 1|2|3|4|5:
        print("weekday")
    case 6|7:
        print("weekend")
    case _:
        print("enter correct value")
```
