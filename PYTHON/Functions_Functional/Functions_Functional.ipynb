```python
# default arguments
# default arguments should be given at last
def greet(name,greeting="Hello"):
    return f"{greeting} {name}"
print(greet("mahesh")) # here greeting by default takes hello when we dont pass as argument
print(greet("Mahesh","Hi")) # here greeting argument will be overridden

# TRAP using mutuable default as parameters
def add_item(item,lst=[]): # list creation happens here 
    lst.append(item)
    return lst
print(add_item(1))
print(add_item(2))
print(add_item(3))

# SAFE better to use immutable defaults 
def add_item(item,lst=None): # list creation happens here 
    if lst is None:
        lst=[]
    lst.append(item)
    return lst
        
print(add_item(1))
print(add_item(2))
print(add_item(3))

import datetime

def log(msg, timestamp=datetime.datetime.now()):   # timestand will not change
    print(f"[{timestamp}] {msg}")
log("mahesh")
log("Ho")

# now it will change    
def log(msg, timestamp=None):
    if timestamp is None:
        timestamp = datetime.datetime.now()
    print(f"[{timestamp}] {msg}")

log("mahesh")
log("Ho")

# mutable defaults for some cases if we want already executed result 
def fib(n,cache={}):
    if n in cache:
        return cache[n]
    cache[n]=n if n<2 else fib(n-1)+fib(n-2)
    return cache[n]

print(fib(6))
print(fib(6))
print(fib(5))
print(fib(-3))
```

```python
# positional arguments
def add(a,b): # here a,b are positional arguments ,order matter
    print(a+b)
add(5,4) 

# keyword arguments
add(a=60,b=89) # here a,b are keyword arguments,order does not matter
```

```python
# *args
# collects extra positional arguments
# if the argument size is unknown we use * args to unpack into tuple of parameters
def add(*args):
    print(sum(args))
    print(type(args))
add(1,2,3,5)
add(1,2,3,4,5,6,7)

# mixing with normal
def add(a,*args):
    print(f"{a} is normal argument")
    print(sum(args))
add(1,2,3,4,54,6) #here a is 1
```

```python
# **kwArgs
# collects extra keyword arguments
def info(**kwArgs):
    for k,v in kwArgs.items():
        print(f"{k}:{v}")
    print(type(kwArgs))
info(name="mahesh",age=25,school="zphs")
```

```python
# mix *args and ** kwArgs
# everything in between should be keyword argument
def info(*args,b,**kwArgs):
    print(f"positional arguments are {args}")
    print(f"keyword argument is {b}")
    print(f"keyword arguments are {kwArgs}")
info(1,2,3,4,5,b=8,name="mahesh",age=25,school="zphs")
```

```python
# scope determines where a variable can be seen and  where to use

#Local scope
# values defined inside function
def Add():
    a=10
    return a+a
print(Add())

#Global scope
# Values defined at top of the script or module
x=80
def Add():
    return x+x
print(Add())

#Enclosing scope
#defined inside outer loop and outside inner loop and used in inner loop
def Add():
    y=90
    def Sub():
        print(y)
    return Sub()
Add()

# built in scope
#print,len,range etc
print(len([1,2,4,5,6,7]))
```

```python
# closure is a function that remembers variables from the scope it was created in (enclosing scope) even after outer loop completes
def Add():
    x=5
    def Sub():
        print(x)
    return Sub
result=Add()
result() # x is still accesible even after outer loop  already returned
#print(x) # this will throw error ,x is not global , x will be in pocket of Sub function
```

```python
# lambda functions
# one line expressions
# do not use statements or multiple lines
x=lambda s:s**2
print(x(6))

sum=lambda a,b:a+b
print(sum(5,7))

greet =lambda name:f"hello {name}!"
print(greet("Mahesh"))

check=lambda x:x%2==0
print(check(8))


# sorted()
# wrt length it will be sorted
words=["jinka","mahesh","r"]
print(sorted(words,key=lambda x:len(x)))


# map()
# applying function to every item
mapping=map(lambda x:x**2,[1,2,3,4])
print(list(mapping))

# multiple iterables
mapping=map(lambda x,y:x+y,[1,2,3,4],[5,6,7,8])
print(list(mapping))


# filter
# keeps items where function returns truthy
nums=[1,2,3,4,5,6,7]
fil=filter(lambda x:x%2==0,nums)
print(list(fil))


# None 
# function filters out falsy items
values = [1, 0, "hello", "", None, [1], []]
fil=filter(None,values)
print(list(fil))


# reduce
# collapse sequence to one value
lst=[1,2,3,4]
from functools import reduce
result=reduce(lambda a,b:a+b,lst)
print(result)
```

```python
# sorted() vs sort()
# sort() operated only on list,which will return None
# sorted() operates with any type,original will be unchanged and creates new one

# sort()
lst=[1,2,3,4]
l=lst.sort()
print(l)
print(lst)

# sorted()
lst=[5,4,3,2,1]
l=sorted(lst) # here sorted() returns
print(l)

print(sorted("mahesh")) # returns as list
```

```python
# key
# key takes a function and calls it on each item
words=["jinka","mahesh","reddy"]
print(sorted(words))
print(sorted(words,key=len))
```

```python
# common key patterns
 

# case -insensitive
names=["cherry","banana","apple","Dog"]
print(sorted(names)) # capitals sorted first
print(sorted(names,key=str.lower)) # alphabetical sort

# tuple elements
t=(("mahesh",25),("asritha",50),("None",15))
print(sorted(t,key=lambda x:x[1]))


# dict elements
d=dict(a=1,b=2,c=3,d=4,e=1)
print(sorted(d.items(),key=lambda x:x[1]))


# list of dict
products=[{"mahesh":2,"score":99},{"asritha":5,"score":100}]
print(sorted(products,key=lambda x:x["score"]))


# object attribute
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

people=[Person("mahesh",100),Person("Asritha",99)]
print(sorted(people,key=lambda x:x.age))
```

```python
# itemgetter,attrgetter
# safer instead of using lambda functions
from operator import itemgetter,attrgetter

class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

people=[Person("mahesh",100),Person("Asritha",99)]

print(sorted(people,key=attrgetter("age")))

t=(("mahesh",25),("asritha",50),("None",15))
print(sorted(t,key=itemgetter(1)))
```

```python
# functools

from functools import partial

# partial
# if we know some of the arguments are same everytime we can lock them in partial and can pass other arguments when calling a fucntion
# partial returns a new function that can be called later
# partial is faster than lambda
def power(base,exponent):
    print(base**exponent)
# if we want to try with the same exponent
square=partial(power,exponent=2)
print(square(5))
print(square(10))


# 2 ways to lock arguments
# 1.positional 
def greet(greeting,name):
    print(f"{greeting} {name}")
g=partial(greet,"hello") # here hello goes to 1st argument which is greeting
g("mahesh")
g("asritha")

# 2.keyword
# which is base exponent example
```

```python
# lru_cache
# it remembers what the function returns
# arguments must be hashable

# the problem
def fib(n):
    if n<2:
        return n
    return fib(n-1)+fib(n-2)
#Slow because it computes the same values repeatedly.

# fix
from functools import lru_cache

@lru_cache(maxsize=None) # remembers everything,we can use @cache also for remembering everything
def fib(n):
    if n<2:
        return n
    return fib(n-1)+fib(n-2)
print(fib(99))
# decorator keeps dictionary behind the scenes
# if value present in dictionary return,else run and save answer

@lru_cache(maxsize=128) # remembers last 128 last only
def fib(n):
    if n<2:
        return n
    return fib(n-1)+fib(n-2)

print(fib(90))
```
